```python
#if-else condition
age = int(input())
if age>=18:    #if condition
  print("you can vote")
else:         #else condition
  print("u cannot vote")
```

```python
#write a program for weeek days
n = int(input("Enter a number: "))
if n==0:
  print("sunday")
elif n==1:
  print("monday")
elif n==2:
  print("tuesday")
elif n==3:
  print("wenesday")
elif n==4:
  print("thruesday")
elif n==5:
  print("friday")
elif n==6:
  print("saturday")
else:
  print("invalid")
```

```python
#check the greatest of two
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

if a > b:
    print("Greatest number is a")
elif b > a:
    print("Greatest number is b")
else:
  print("both are same")
```
